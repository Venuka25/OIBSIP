```python
# Create a copy for cleaning
df_clean = df.copy()

# Store before-cleaning information
before_rows = len(df_clean)
before_missing = df_clean.isnull().sum().sum()
before_duplicates = df_clean.duplicated().sum()

print("Original Rows:", before_rows)
print("Missing Values:", before_missing)
print("Duplicate Rows:", before_duplicates)
```

### Missing Value Handling

1. Age: Missing values are replaced with the median because it is less affected by outliers.
2. Embarked: Missing values are replaced with the mode because it is a categorical column.
3. Cabin: This column is removed because it contains a large number of missing values.

```python
# Fill missing Age values with median
df_clean["Age"] = df_clean["Age"].fillna(df_clean["Age"].median())

# Fill missing Embarked values with mode
df_clean["Embarked"] = df_clean["Embarked"].fillna(
    df_clean["Embarked"].mode()[0]
)

# Drop Cabin column
df_clean.drop(columns=["Cabin"], inplace=True)

print("Missing values handled successfully!")

print("\nRemaining Missing Values:")
print(df_clean.isnull().sum())
```

```python
# Count duplicates before removal
duplicates_before = df_clean.duplicated().sum()

# Remove duplicate rows
df_clean.drop_duplicates(inplace=True)

print("Duplicates Removed:", duplicates_before)
print("Rows After Removing Duplicates:", len(df_clean))
```

### Data Standardization

Categorical values are standardized to avoid inconsistent formats.

- Sex: Converted to lowercase.
- Embarked: Converted to uppercase.
- Extra spaces are removed from text columns.

```python
# Remove extra spaces from text columns
for col in df_clean.select_dtypes(include="object").columns:
    df_clean[col] = df_clean[col].str.strip()

# Standardize Sex
df_clean["Sex"] = df_clean["Sex"].str.lower()

# Standardize Embarked
df_clean["Embarked"] = df_clean["Embarked"].str.upper()

print("Standardization completed!")

print("\nSex Values:")
print(df_clean["Sex"].unique())

print("\nEmbarked Values:")
print(df_clean["Embarked"].unique())
```

```python
# Convert PassengerId and Ticket to string
df_clean["PassengerId"] = df_clean["PassengerId"].astype(str)
df_clean["Ticket"] = df_clean["Ticket"].astype(str)

# Ensure numerical columns have correct types
df_clean["Age"] = pd.to_numeric(df_clean["Age"])
df_clean["Fare"] = pd.to_numeric(df_clean["Fare"])

print("Data types corrected!")
print(df_clean.dtypes)
```

### Outlier Detection Using IQR Method

The IQR method is used to identify extreme values.

- Age: Extreme values are retained because they may represent genuine passenger ages.
- Fare: High values are retained because they may represent expensive tickets.
- SibSp and Parch: Extreme counts are retained because they represent actual family sizes.

No valid observations are removed or capped.

```python
# Detect outliers using IQR method

outlier_columns = ["Age", "Fare", "SibSp", "Parch"]

for col in outlier_columns:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[col] < lower) | (df_clean[col] > upper)
    ]

    print(col, "Outliers:", len(outliers))

print("\nOutlier detection completed!")
```

```python
print("Cleaned Dataset Shape:", df_clean.shape)

print("\nRemaining Missing Values:")
print(df_clean.isnull().sum())

print("\nRemaining Duplicate Rows:")
print(df_clean.duplicated().sum())

print("\nFinal Data Types:")
print(df_clean.dtypes)
```

```python
# Create before and after comparison

comparison = pd.DataFrame({
    "Metric": [
        "Total Rows",
        "Total Columns",
        "Missing Values",
        "Duplicate Rows"
    ],
    "Before Cleaning": [
        len(df),
        len(df.columns),
        df.isnull().sum().sum(),
        df.duplicated().sum()
    ],
    "After Cleaning": [
        len(df_clean),
        len(df_clean.columns),
        df_clean.isnull().sum().sum(),
        df_clean.duplicated().sum()
    ]
})

display(comparison)
```
