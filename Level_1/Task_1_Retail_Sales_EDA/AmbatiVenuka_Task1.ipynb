```python
# Convert Date column to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Create Month column
df['Month'] = df['Date'].dt.month_name()

# Check the result
print(df[['Date', 'Month']].head())
```

```python
month_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

df['Month'] = pd.Categorical(
    df['Month'],
    categories=month_order,
    ordered=True
)

monthly_sales = df.groupby(
    'Month',
    observed=False
)['Total Amount'].sum()

print(monthly_sales)
```

```python
plt.figure(figsize=(10, 5))

plt.bar(monthly_sales.index, monthly_sales.values)

plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Total Sales Amount')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
monthly_average = df.groupby(
    'Month',
    observed=False
)['Total Amount'].mean()

print(monthly_average)
```

```python
plt.figure(figsize=(10, 5))

plt.bar(monthly_average.index, monthly_average.values)

plt.title('Average Sales Amount per Month')
plt.xlabel('Month')
plt.ylabel('Average Total Amount')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker='o'
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales Amount')
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Create Quarter column
df['Quarter'] = df['Date'].dt.to_period('Q').astype(str)

# Calculate quarterly sales
quarterly_sales = df.groupby('Quarter')['Total Amount'].sum()

print(quarterly_sales)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    quarterly_sales.index,
    quarterly_sales.values,
    marker='o'
)

plt.title('Quarterly Sales Trend')
plt.xlabel('Quarter')
plt.ylabel('Total Sales Amount')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Correlation Heatmap

import seaborn as sns

numeric_columns = [
    'Age',
    'Quantity',
    'Price per Unit',
    'Total Amount'
]

correlation_matrix = df[numeric_columns].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Matrix of Numerical Variables')
plt.tight_layout()
plt.show()
```

## Key Observations

### Monthly Sales Trend
Monthly sales vary throughout the year, showing changes in customer purchasing activity across different months.

### Quarterly Sales Trend
Quarterly sales provide a broader view of the yearly sales pattern and help identify periods with relatively higher or lower revenue.

### Product Category Analysis
Sales and quantity analysis across product categories shows differences in category-level performance and customer demand.

### Gender Analysis
Sales, quantity purchased, and average spending were compared across genders to identify differences in purchasing behaviour.

### Age Group Analysis
The age-group analysis shows how sales and transaction activity are distributed across different customer age ranges.

### Age vs. Total Amount
The scatter plot helps examine the relationship between customer age and transaction amount. The points show that transaction values vary across different ages.

### Correlation Heatmap
The correlation matrix shows the strength and direction of relationships among the numerical variables, including Age, Quantity, Price per Unit, and Total Amount.

## Product-Level Analysis Limitation

The dataset does not contain a Product Name or Product ID column. Therefore, a Top 10 Best-Selling Products analysis cannot be performed directly. Product Category was used instead to analyse product-level sales performance.

## Actionable Business Recommendations

1. Monitor monthly and quarterly sales trends to identify periods of higher and lower demand and plan inventory accordingly.

2. Focus inventory and promotional planning on product categories with stronger sales and quantity performance, while investigating weaker categories for potential improvement.

3. Use customer age-group and gender purchasing patterns to develop more targeted marketing campaigns and product promotions.

4. Track average transaction value alongside total sales so that the business can identify opportunities to increase customer spending per transaction.

5. Collect Product Name or Product ID information in future datasets to enable detailed product-level analysis and identification of the top-selling products.

## Final EDA Findings and Conclusion

The Retail Sales dataset contains 1,000 transactions with 11 columns and no missing values.

The total sales amount is 456,000, with an average transaction amount of 456. A total of 2,514 units were sold across 1,000 transactions.

Monthly sales analysis was performed to understand sales trends throughout the year. Product category analysis was used to compare sales and quantities across different categories.

Gender-based analysis was performed to compare total sales, quantity sold, and average spending between male and female customers.

Age group analysis was conducted to understand sales distribution and transaction counts across different customer age groups. A scatter plot was also used to examine the relationship between customer age and total transaction amount.

Overall, the exploratory data analysis provides an understanding of sales trends, customer characteristics, product category performance, and purchasing patterns in the retail sales dataset.
